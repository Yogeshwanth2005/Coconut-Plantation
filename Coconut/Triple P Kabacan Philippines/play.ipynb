```python
import xml.etree.ElementTree as ET
import json
import os

def cvat_xml_to_multiple_geojson(xml_path, output_prefix="Triple p"):
    tree = ET.parse(xml_path)
    root = tree.getroot()

    image_tags = root.findall('image')
    
    for idx, image_tag in enumerate(image_tags, start=1):
        image_name = image_tag.attrib['name']
        features = []

        for box in image_tag.findall('box'):
            label = box.attrib['label']
            xtl = float(box.attrib['xtl'])
            ytl = float(box.attrib['ytl'])
            xbr = float(box.attrib['xbr'])
            ybr = float(box.attrib['ybr'])

            coords = [
                [xtl, ytl],
                [xbr, ytl],
                [xbr, ybr],
                [xtl, ybr],
                [xtl, ytl]
            ]

            feature = {
                "type": "Feature",
                "properties": {
                    "classification": label,
                    "image": image_name
                },
                "geometry": {
                    "type": "Polygon",
                    "coordinates": [coords]
                }
            }

            features.append(feature)

        geojson = {
            "type": "FeatureCollection",
            "features": features
        }

        output_filename = f"{output_prefix}{idx}.geojson"
        with open(output_filename, "w") as f:
            json.dump(geojson, f, indent=2)
        print(f"Saved {output_filename} for image: {image_name}")

# Example usage
cvat_xml_to_multiple_geojson("t3.xml")
```
