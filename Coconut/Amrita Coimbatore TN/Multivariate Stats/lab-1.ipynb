```python
import pandas as pd

# Load the CSV file
df = pd.read_csv("Air_Pollution_Data.csv")

# Select valid numeric columns
data = df[['Wind (x1)', 'Solar radiation (x2)', 'CO (x3)', 'NO (x4)', 'NO2 (x5)', 'O3 (x6)', 'HC (x7)']]

# Compute population statistics (ddof=0)
average = data.mean()
variance = data.var(ddof=0)
std_dev = data.std(ddof=0)

# Compute population covariance matrix manually
def population_cov(df):
    centered = df - df.mean()
    return (centered.T @ centered) / len(df)

covariance = population_cov(data)

# Display results
print("Population Mean:\n", average)
print("\nPopulation Variance:\n", variance)
print("\nPopulation Standard Deviation:\n", std_dev)
print("\nPopulation Covariance Matrix:\n", covariance)
```
