```python
import pandas as pd
df = pd.read_excel("aircraft_question.xlsx")
data = df[['y', 'x1', 'x2', 'x3']]
average = data.mean()
variance = data.var(ddof=0)
std_dev = data.std(ddof=0)
def population_cov(df):
    centered = df - df.mean()
    return (centered.T @ centered) / len(df)
covariance = population_cov(data)
print("Mean:", average)
print("Variance:", variance)
print("Standard Deviation:", std_dev)
print("Covariance Matrix:", covariance)
```
