```python
import pandas as pd

file_path = "Air_Pollution_Data.csv"
df = pd.read_csv(file_path)

numeric_df = df.select_dtypes(include='number').astype(float)

print("mean:", numeric_df.mean())
print("variance:", numeric_df.var(ddof=0))
print("standard deviation:", numeric_df.std(ddof=0))
print("covariance:", numeric_df.cov(ddof=0))
```
